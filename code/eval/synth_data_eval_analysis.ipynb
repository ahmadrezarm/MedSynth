```python
import matplotlib.pyplot as plt
import numpy as np

# Data preparation
samples = sorted(results_df['Samples'].unique())
datasets = ['Aci_Train', 'NoteChat+Aci_Train', 'MedSynth+Aci_Train', "NoteChat", "MedSynth"]


colors = {
    'Aci_Train': 'orange',
    'NoteChat+Aci_Train': 'red',
    'MedSynth+Aci_Train': 'blue',
    "NoteChat": "pink",
    "MedSynth":"purple"
}

# Assigning colors based on the Dataset
results_df['Color'] = results_df['Dataset'].map(colors)

# Create a pivot table to organize the data
pivot_table = results_df.pivot(index='Samples', columns='Dataset', values='avg_score').reindex(samples, fill_value=0)

x = np.arange(len(samples))  # the label locations
width = 0.3  # the width of the bars
multiplier = 0

fig, ax = plt.subplots(figsize=(14, 8))

# Plot each dataset as a group of bars
for dataset in datasets:
    offset = width * multiplier
    rects = ax.bar(x + offset, pivot_table[dataset], width, label=dataset, color=colors[dataset])
    #ax.bar_label(rects, padding=3)
    multiplier += 1

# Add labels, title, and custom x-axis tick labels
ax.set_xlabel('Sample Size')
ax.set_ylabel('Average Score')
ax.set_title('Average Score by Dataset and Sample Size')
ax.set_xticks(x + width, samples)
ax.legend(title="Datasets")
ax.set_ylim(0, max(results_df['avg_score']) * 1.1)

plt.tight_layout()
plt.show()
```
