```python
import evaluate
import numpy as np

class MetricsComputer:
    def __init__(self, prediction_list, gt_list):
        self.pred_list = prediction_list
        self.gt_list = gt_list

    def compute_BLEU(self):
        bleu = evaluate.load("bleu")
        bleu_scores = []

        for pred, ref in zip(self.pred_list, self.gt_list):
            score = bleu.compute(predictions=[pred], references=[ref])["bleu"]
            bleu_scores.append(score)

        return {"mean": np.mean(bleu_scores), "std": np.std(bleu_scores)}

    def compute_ROUGE(self):
        rouge = evaluate.load('rouge')
        rouge_scores = {
            "rouge1": [], "rouge2": [], "rougeL": [], "rougeLsum": []
        }

        for pred, ref in zip(self.pred_list, self.gt_list):
            result = rouge.compute(predictions=[pred], references=[ref])
            rouge_scores["rouge1"].append(result["rouge1"])
            rouge_scores["rouge2"].append(result["rouge2"])
            rouge_scores["rougeL"].append(result["rougeL"])
            rouge_scores["rougeLsum"].append(result["rougeLsum"])  

        return {
            "rouge1": {"mean": np.mean(rouge_scores["rouge1"]), "std": np.std(rouge_scores["rouge1"])},
            "rouge2": {"mean": np.mean(rouge_scores["rouge2"]), "std": np.std(rouge_scores["rouge2"])},
            "rougeL": {"mean": np.mean(rouge_scores["rougeL"]), "std": np.std(rouge_scores["rougeL"])},
            "rougeLsum": {"mean": np.mean(rouge_scores["rougeLsum"]), "std": np.std(rouge_scores["rougeLsum"])}
        }

    # def compute_BERTScore(self):
    #     bertscore = evaluate.load("bertscore")
    #     results = bertscore.compute(predictions=self.pred_list, references=self.gt_list, lang="en")
    #     return {"mean": np.mean(results['f1'])}  # No std for BERTScore

    def compute_METEOR(self):
        meteor = evaluate.load('meteor')
        meteor_scores = []

        for pred, ref in zip(self.pred_list, self.gt_list):
            score = meteor.compute(predictions=[pred], references=[ref])["meteor"]
            meteor_scores.append(score)

        return {"mean": np.mean(meteor_scores), "std": np.std(meteor_scores)}
```
