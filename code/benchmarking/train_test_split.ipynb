```python
import pandas as pd
from sklearn.model_selection import train_test_split

our_data= pd.read_csv("/h/ahmad/SynthDataGen/Synthetic_Data_Gen/data/output/notes_onVector/phase_2_to_12_combined/total_data.csv", sep="|")
filtered_df= our_data[["Disease Description", "Polished Note", "Role", "polished_dial"]]

icd10_df= pd.read_csv("/h/ahmad/SynthDataGen/Synthetic_Data_Gen/data/input/IQVIA/IQVIA_cleaned.csv", sep= "|")

merged_df = filtered_df.merge(icd10_df[['ICD10_desc', 'ICD10']], 
                              left_on='Disease Description', 
                              right_on='ICD10_desc', 
                              how='left')

# Drop the auxiliary 'ICD10_desc' column
merged_df.drop('ICD10_desc', axis=1, inplace=True)

# Drop duplicates to ensure no extra rows are added
merged_df = merged_df.drop_duplicates(subset=['Disease Description', 'Role', 'Polished Note', 'polished_dial'])

# Reorder the columns as per the required order
reordered_df = merged_df[['ICD10', 'Disease Description', 'Role', 'Polished Note', 'polished_dial']]

```

```python
train_df, test_df = train_test_split(reordered_df, test_size=0.15, random_state=96)

# Optionally, you can reset the index of both DataFrames if needed
train_df = train_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)


print(len(train_df))
print(len(test_df))
```

```python
train_df.to_csv("/h/ahmad/SynthDataGen/Synthetic_Data_Gen/code/benchmarking/dataset/train.csv", sep="|")
test_df.to_csv("/h/ahmad/SynthDataGen/Synthetic_Data_Gen/code/benchmarking/dataset/test.csv", sep= "|")
```
